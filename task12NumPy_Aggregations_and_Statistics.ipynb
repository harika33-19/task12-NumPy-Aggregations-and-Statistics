```python
import numpy as np

print("NumPy version:", np.__version__)
print("NumPy imported successfully!")
```

```python
marks = np.array([
    [85, 90, 88],
    [78, 82, 80],
    [92, 95, 94],
    [70, 75, 72],
    [88, 86, 91]
])
print("Student Marks Dataset:")
print(marks)
```

```python
print("Total marks:", np.sum(marks))
print("Mean:", np.mean(marks))
print("Median:", np.median(marks))
print("Minimum:", np.min(marks))
print("Maximum:", np.max(marks))
print("Standard Deviation:", np.std(marks))
print("Variance:", np.var(marks))
```

```python
print("Subject-wise Total:", np.sum(marks, axis=0))
print("Subject-wise Mean:", np.mean(marks, axis=0))
print("Subject-wise Median:", np.median(marks, axis=0))
print("Subject-wise Minimum:", np.min(marks, axis=0))
print("Subject-wise Maximum:", np.max(marks, axis=0))
print("Subject-wise Standard Deviation:", np.std(marks, axis=0))
print("Subject-wise Variance:", np.var(marks, axis=0))
```

```python
print("Student-wise Total:", np.sum(marks, axis=1))
print("Student-wise Mean:", np.mean(marks, axis=1))
print("Student-wise Median:", np.median(marks, axis=1))
print("Student-wise Minimum:", np.min(marks, axis=1))
print("Student-wise Maximum:", np.max(marks, axis=1))
print("Student-wise Standard Deviation:", np.std(marks, axis=1))
print("Student-wise Variance:", np.var(marks, axis=1))
```

```python
subjects = ["Python", "Mathematics", "Statistics"]

subject_means = np.mean(marks, axis=0)
subject_stds = np.std(marks, axis=0)
subject_vars = np.var(marks, axis=0)

for i in range(len(subjects)):
    print("\nSubject:", subjects[i])
    print("Mean:", round(subject_means[i], 2))
    print("Standard Deviation:", round(subject_stds[i], 2))
    print("Variance:", round(subject_vars[i], 2))
```

```python

student_totals = np.sum(marks, axis=1)

highest_student = np.argmax(student_totals) + 1
lowest_student = np.argmin(student_totals) + 1

print("Student with highest total:", highest_student)
print("Highest total marks:", np.max(student_totals))

print("Student with lowest total:", lowest_student)
print("Lowest total marks:", np.min(student_totals))
```
